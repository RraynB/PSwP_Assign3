```python
#initial building the game with random movements

import numpy as np

def play_nim(n,s): 
    heaps = [0]*n
    for i in range(s): 
        chosen_heap = np.random.randint(0,n)
        heaps[chosen_heap] += 1
    turn = 1 
    while len(heaps) != 0: 
        pos = np.random.randint(0,len(heaps)) 
        stones = np.random.randint(1, heaps[pos]+1)
        heaps[pos] -= stones
        if heaps[pos] == 0: 
            heaps.pop(pos)
        turn += 1
        if turn % 2 == 1: 
            player = "Player 1" 
        else: 
            player = "Player 2"
            
        print(player, heaps)

    return(player, "loses")

print(play_nim(3,10)) 

```

```python
def nim_sum(heaps):
    binary = [bin(h)[2:] for h in heaps]
    binary = [int(h, 2) for h in binary]

    result = 0

    for h in binary:
        result ^= h

    return int(bin(result), 2)

def nim_winner(heaps):

    if max(heaps) == 1:
        if len(heaps) % 2 == 1:
            return False
        else:
            return True 

    if nim_sum(heaps) == 0:
        return False 
    else: 
        return True 

def phase1_p1_wins(n, s, require_nonempty=False):

    def mover_wins(heaps, remaining):
        if remaining == 0:
            return nim_winner(heaps)
        for h in set(heaps):
            new = list(heaps); new.remove(h); new.append(h + 1)
            new = tuple(sorted(new))
            if remaining - 1 < new.count(0):
                continue                     
            if not mover_wins(new, remaining - 1):
                return True
        return False

    return mover_wins((0,) * n, s)

for n in range(1, 11):
    print(f"n={n}:", "".join("W" if phase1_p1_wins(n, s) else "L"
                             for s in range(n, 21)))

    
```
